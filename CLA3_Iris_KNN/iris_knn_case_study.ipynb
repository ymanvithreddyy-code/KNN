# Iris Flower Species Classification Using KNN

A compact multi-class classification case study using Scikit-learn's built-in Iris dataset.

## 1. Load data and split it
The dataset contains 150 samples, four numeric measurements, and three species labels. We use a stratified 80:20 train-test split.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

iris = load_iris()
X, y = iris.data, iris.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
```

## 2. Train the KNN model
StandardScaler puts the four measurements on comparable scales. KNN with k=5 predicts the class by majority vote among the five nearest training samples.

```python
model = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5))
model.fit(X_train, y_train)
predictions = model.predict(X_test)
```

## 3. Evaluate the model

```python
print("Accuracy:", round(accuracy_score(y_test, predictions) * 100, 2), "%")
print("\nConfusion Matrix:\n", confusion_matrix(y_test, predictions))
print("\nClassification Report:\n",
      classification_report(y_test, predictions, target_names=iris.target_names))
```

## 4. Predict a new flower
Enter measurements in this order: sepal length, sepal width, petal length, petal width (cm).

```python
sample = [[5.1, 3.5, 1.4, 0.2]]
result = model.predict(sample)[0]
print("Predicted species:", iris.target_names[result])
```
